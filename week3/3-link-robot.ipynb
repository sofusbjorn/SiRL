## Task 1: Loading and interpreting a threelink model

```python
import mujoco
import mujoco.viewer
import time
from pathlib import Path

# Load model from XML files
model = mujoco.MjModel.from_xml_path("threelinks.xml")
```

```python
"""
• What are the dimensions of the link?
• How are the links connected to each other?
• What is the furthest distance that the ”arm tip” can reach?
"""

# Get the dimensions of the links
print(model.geom_size)
print(model.geom_type)
print("Link 1 dimensions:", model.geom_size[0])  # Assuming the first geom is link 1
print("Link 2 dimensions:", model.geom_size[1])  # Assuming the second geom is link 2
print("Link 3 dimensions:", model.geom_size[2])  # Assuming the third geom is link 3


# Get the connections between the links
print("Joint types:", model.joint)
```

```python

# Launch a viewer
with mujoco.viewer.launch(model) as viewer:
    while viewer.is_running():
        print("Updating model from robot...", flush=True)
        # Update the model's joint positions from the robot
        viewer.sync()
        time.sleep(0.01)
```
